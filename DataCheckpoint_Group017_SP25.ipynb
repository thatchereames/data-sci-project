## Background and Prior Work

The relationship between weather and human well-being has been a topic of interest in environmental research for years, with strong connections to psychological studies. Numerous studies have shown that weather conditions such as sunny days tend to improve mood, while rainy days can increase feelings of sadness. Although these findings are valuable, weather is a complex phenomenon that encompasses a wide range of factors, making it challenging to grasp its direct effects on human well-being. To address this, our research aims to investigate the role of temperature variability within a single day. We will examine whether the fluctuation between daytime high and nighttime low temperatures has a profound effect on a city’s overall happiness. By focusing on temperature variability rather than climate trends, we can more accurately identify patterns that reveal how weather affects happiness levels.

Past research has utilized long-term average temperatures to draw conclusions about the impact of weather on happiness. For example, Rehdanz’s study of climate trends in 67 countries, using mean temperature and precipitation in the hottest and coldest months, found that higher temperatures during the coldest month were associated with greater happiness.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Another study by Tsutsui found that happiness was highest at a maximum temperature of 13.9 Celsius.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) However, the results also indicated that although temperature influences people’s happiness, there is considerable variation in their responses. Few studies have explored how day-to-day weather changes, such as shifts between extreme and low temperatures, affect individuals’ happiness. A notable exception is research by Lucas and Lawless, which analyzed data from 1 million Americans over five years, combining different daily weather variables with life satisfaction ratings. They concluded that there was no significant correlation between temperature and life satisfaction judgments.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

**Our work narrows the focus to temperature variability and its relationship with happiness at the city level. Past work has tendened to focus on average temperatures and its relationship to happiness. Does the overall consistency of the weather, consistently hot or cold rather than prone to large fluctutions throughout the year, correlate with overall happiness. Does stability bring contentment and peace?**

This question connects to a broader idea in psychology: the more uncontrolled factors in your life, the more likely they are to impact your mental health. Weather, as an external and often unpredictable force, may function as one of these factors. If we view stability as a source of comfort and predictability, then cities with more stable weather patterns might foster a greater sense of peace and happiness among residents. Constant change, even in something as mundane as temperature, could contribute to stress or discomfort. Does greater stability in the environment support mental well-being? This is the central question our project seeks to explore.

The Deon's Checklist of Ethics as it pertains to our project currently:

#### Are there any biases/privacy/terms of use issues with the data you proposed?

There are currently no known privacy or terms of use issues with our datasets. The World Happiness Report and NOAA’s Global Historical Climatology Network (GHCN) are both publicly available, well-documented, and widely used in academic research. However, we recognize that the World Happiness Report does not always disclose how each country’s data was collected in detail. This raises the possibility of inconsistent survey methods, informed consent standards, or sampling coverage, especially in countries with limited transparency or different data regulations. We will take these concerns into account during our analysis and interpretation.


#### Are there potential biases in your dataset(s), in terms of who it composes, and how it was collected, that may be problematic in terms of it allowing for equitable analysis?


Yes — both datasets carry potential biases that may limit how equitably we can analyze happiness across cities:
The World Happiness Report surveys (from Gallup World Polls) tend to sample individuals with access to stable internet, literacy, and willingness to participate in long surveys. As a result, marginalized groups — such as low-income individuals, non-native speakers, or people in informal housing — may be underrepresented. The report also aggregates data at the national level, and while we will map scores to cities, this may obscure within-country urban vs rural differences.

The NOAA GHCN temperature data is heavily reliant on official weather stations, which are more densely placed in developed, urban areas. Cities in the Global South or rural towns may have missing or sparse data, skewing the geographic representativeness of our dataset. This could create an urban bias in which only well-monitored cities are analyzed. Although NOAA’s climate data is considered reliable, it still reflects geographic and resource-based disparities in monitoring between urban and rural communities.
Results could also be misinterpreted where we think there is a direct link between temperature and happiness data but we are not accounting for other confounding variables, such as income level.


#### How will you set out to detect these specific biases before, during, and after/when communicating your analysis?

**Before**: We will read through the documentation for both datasets fully to identify known limitations in data collection, such as underrepresented regions or demographic groups, and also to get a comprehensive understanding of our datasets. We’ll also assess how happiness surveys were conducted (e.g., sampling frame, language accessibility, survey mode) and note any missing data in both datasets (climate records and happiness), especially from less-monitored regions. 

**During**: We’ll perform exploratory data analysis to identify outliers, data gaps, or irregular patterns that may signal bias — such as cities with incomplete climate data or abnormally high happiness scores in low-income regions. We’ll also test how sensitive our results are to the exclusion of certain cities or regions (sensitivity analysis).
After: In our final report and presentations, we will include clear sections that explain the potential biases, risks of misinterpretation, and overall limitations. We’ll avoid overgeneralizing or implying causation and will be transparent about where our data coverage is weaker.

#### Are there any other issues related to your topic area, data, and/or analyses that are potentially problematic in terms of data privacy and equitable impact? How will you handle issues you identified?

The datasets do not have privacy concerns (specifically concerning identifiable personal data), but there are other potential problem areas. 
Measuring happiness can be really subjective and have different cultural implications → Relying on a single numerical score for each city may ignore differences across subpopulations — like low-income residents, immigrants, or people experiencing housing insecurity. Those are all serious effectors of happiness that do not get accounted for.
Both datasets are more accurate measures for urban areas because of the imbalance in the amount of data collected in more rural areas. 

We also have to be careful not to make generalized statements like, “warmer cities are happier”, because there are other confounding variables such as income level. This is really important because if misinterpreted by policymakers a situation where a politician reallocates city funds toward heat mitigation or green spaces — assuming that temperature alone drives happiness — could occur. They would be ignoring the many other factors that affect an individual’s happiness. For example, the root causes of unhappiness in low-income communities might not actually be about temperature and lead to misguided or inequitable interventions.

We will handle all of these issues by clearly acknowledging the limits of our data and how it affects our project. Additionally, we will frame our analysis in terms of correlations rather than causation, and we’ll avoid using language that implies direct or universal relationships between variables.
